```python
import pandas as pd

# Load the dataset
df = pd.read_json("English_Olympiad_Mathematics.json")

# Filter rows where label == "Theorem_Proving" AND 'ARML' is in the source
filtered = df[
    (df["label"] == "Theorem_Proving") &
    (df["source"].str.contains("ARML", case=False, na=False))
]

# Show the matching rows
print(filtered)

# (Optional) show only relevant columns for clarity
print(filtered[["source", "label", "question", "solution"]])
```

```python
import pandas as pd

# Load dataset
df = pd.read_json("English_Olympiad_Physics.json")

# Filter only IPhO-related rows
filtered = df[df["source"].str.contains("IPhO", case=False, na=False)].copy()

# Extract the year (e.g., "IPhO/2015" → "2015")
filtered["year"] = filtered["source"].str.extract(r"IPhO/(\d{4})")

# Count unique IDs per year
counts = filtered.groupby("year")["id"].nunique().reset_index()

# Sort by year
counts = counts.sort_values("year").reset_index(drop=True)

print(counts)
```
